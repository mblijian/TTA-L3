```python
import random

print("Welcome to LUCKY 21!")

player_hand = []
dealer_hand =[]

# give two cards to player and dealer
for i in range(2):
    player_hand.append(random.randint(1,10))
    dealer_hand.append(random.randint(1,10))

# sort the cards from smallest to largest
player_hand.sort()
dealer_hand.sort()

print(f"Your cards : {player_hand}")
print(f"Your total:{sum(player_hand)}")

while sum(player_hand)< 21:
    choice = input("Hit or stand?: Enter h or s").lower()
    if choice == "h":
        new_card = random.randint(1,10)
        player_hand.append(new_card)
        player_hand.sort()

        print(f"You receive:{new_card}")
        print(f"Your cards : {player_hand}")
        print(f"Your total:{sum(player_hand)}")
    
    elif choice =="s":
        print("You chose to stand.")
        break
    else:
        print("Please enter h or s.")

# compare the dealer hand and player hand
if sum(player_hand)> 21:
    print("You went over 21, you loose.")

else:
    print(f"Dealer's cards:{dealer_hand} ")

    while sum(dealer_hand)<17:
        new_card = random.randint(1,10)
        dealer_hand.append(new_card)
        dealer_hand.sort()

        print(f"Your cards : {dealer_hand}")
        print(f"Your total:{sum(dealer_hand)}")
    # show final result
    player_total = sum(player_hand)
    dealer_total = sum(dealer_hand)

    print("FINAL RESULT!")
    print(f"Your cards: {player_hand}")
    print(f"Your total: {sum(player_hand)}")
    print(f"Dealer cards: {dealer_hand}")
    print(f"Dealer total: {sum(dealer_hand)}")

    # Decide the winner!
    if dealer_total>21:
        print("The dealer went over 21. You win!")
    elif player_total>dealer_total:
        print("The player wins!")
    elif player_total<dealer_total:
        print("The dealer wins!")
    else:
        print("It is a draw!")
```
