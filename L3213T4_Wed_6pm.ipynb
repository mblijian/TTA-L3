2. function library

```python
# 1. pick up funciton
def pick_up():
    import time
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(2)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,0,-70,1000)
    time.sleep(2)
    got.mechanical_clamp_close()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(2)

# 2. put down function
def put_down():
    import time
    got.mechanical_joint_control(90,30,30,1000)
    time.sleep(2)
    got.mechanical_joint_control(90,0,-80,1000)
    time.sleep(2)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(90,30,30,1000)
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(2)
    got.mechanical_clamp_close()
    time.sleep(2)

# 3. move and stop
def move_stop():
    while True:
        Laser1 = got.read_distance_data(21)
        if Laser1> 10:
            got.mecanum_move_speed(0,20)
        else:
            got.mecanum_stop()
            break

# 4. maze escape
def maze_escape():
    import time
    move_forward = 0
    while move_forward<4:
        Laser1 = got.read_distance_data(21)
        if Laser1> 10:
            got.mecanum_move_speed(0,20)
        else:
            got.mecanum_stop()
            time.sleep(0.5)
            got.mecanum_turn_speed_times(3,30,90,2)
            move_forward+=1
            print(f"This is move forward:{move_forward}.")
            
            Laser1 = got.read_distance_data(21)
            if Laser1<10:
                got.mecanum_turn_speed_times(2,30,180,2)
                time.sleep(0.5)

# 5. pick up 100% function
def pick_up_guarantee():
    import time
    move_stop()
    pick_up()
    time.sleep(1)

    Laser1 = got.read_distance_data(21)
    while Laser1<12:
        Laser1 = got.read_distance_data(21)
        if Laser1<5:
            got.mecanum_move_speed_times(1,5,5,1)
            time.sleep(0.5)
            pick_up()
            time.sleep(2)

    got.show_light_rgb_effect(0,255,0,3)
    got.play_audio_tts("I got the token! Let's go through the maze now!", 0, False)

# 6. face recognition
def face_recognition():
    while True:
        face_info = got.get_face_recognition_total_info()
        if not face_info:
            got.mecanum_move_speed(0,20)
        else:
            got.mecanum_stop()
            break
```

```python
import random

player_hand = []
dealer_hand = []

# give two cards to player and dealer
for i in range(2):
    player_hand.append(random.randint(1,10))
    dealer_hand.append(random.randint(1,10))

# sort the cards
player_hand.sort()
dealer_hand.sort()

print(f"Your cards:{player_hand}")
print(f"Your total:{sum(player_hand)}")

while sum(player_hand)<21:
    choice = input("Hit or stand?:Enter h or s.").lower()
    if choice =="h":
        new_card =random.randint(1,10)
        player_hand.append(new_card)
        player_hand.sort()

        print(f"You receive:{new_card}")
        print(f"Your cards:{player_hand}")
        print(f"Your total:{sum(player_hand)}")
    elif choice =="s":
        print("You chose to stand.")
        break
    else:
        print("Please chose either h or s.")

# compare the player hand and the dealer hand
if sum(player_hand)> 21:
    print("You went over 21, you lose.")
else:
    print(f"Dealer's cards:{dealer_hand}")

    while sum(dealer_hand)<17:
        new_card =random.randint(1,10)
        dealer_hand.append(new_card)
        dealer_hand.sort()

        print(f"Dealer cards:{dealer_hand}")
        print(f"Dealer total:{sum(dealer_hand)}")
    #show final result
    player_total = sum(player_hand)
    dealer_total = sum(dealer_hand)

    print("FINAL RESULT!")
    print(f"Your cards:{player_hand}")
    print(f"Your total:{sum(player_hand)}")
    print(f"Dealer cards:{dealer_hand}")
    print(f"Dealer total:{sum(dealer_hand)}")

    #Decide the winner!
    if dealer_total>21:
        print("The dealer went over 21, you win!")
        got.play_audio_tts("The dealer went over 21, you win!",1,False)
    elif player_total>dealer_total:
        print("The player wins!")
        got.play_audio_tts("The player wins!",1,False)
    elif player_total<dealer_total:
        print("The dealer wins!")
        got.play_audio_tts("The dealer wins!",1,False)
    else:
        print("It's a draw!")
        got.play_audio_tts("It's a draw!",1,False)
    
```

## Session 7

1. face recognition

```python
# GPT version turn right and centralize
import time

while True:
    face_info = got.get_face_recognition_total_info()
    if face_info and face_info[0][1] < 340:
        got.mecanum_stop()
        break
    got.mecanum_move_xyz(0, 0, -20)
    time.sleep(0.1)
```

```python
# GPT version turn left and centralize
import time

while True:
    face_info = got.get_face_recognition_total_info()
    if face_info and face_info[0][1] > 300:
        got.mecanum_stop()
        break
    got.mecanum_move_xyz(0, 0, 20)
    time.sleep(0.1)
```

2. diagnoal move

```python
import time
while True:
    face_info = got.get_face_recognition_total_info()
    if face_info:
        face_x = face_info[0][1]
        face_width = face_info[0][4]
        # Stop the vehicle
        if face_width> 70 and 300< face_x < 350:
            got.mecanum_stop()
            print("Let's stop")
            break
        elif face_x> 380:
            got.mecanum_move_xyz(20, 10, 0)
        elif face_x< 300:
            got.mecanum_move_xyz(-20, 10, 0)
        else:
            got.mecanum_move_xyz(0, 10, 0)
    else:
        # No face, move forward
        got.mecanum_move_xyz(0, 10, 0)
    time.sleep(0.01)
```

```python
import time
while True:
    face = got.get_face_recognition_total_info()
    if face:
        x, width = face[0][1], face[0][4]
        if width > 70 and 300 < x < 350:
            got.mecanum_stop()
            print("Let's stop")
            break
        move_x = 20 if x > 380 else -20 if x < 300 else 0
    else:
        move_x = 0
    got.mecanum_move_xyz(move_x, 10, 0)
    time.sleep(0.01)
```
